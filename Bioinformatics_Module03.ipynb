````python
#Given: A file containing at most 1000 lines.
#Return: A file containing all the even-numbered lines from the original file. Assume 1-based numbering of lines.

with open("rosalind_ini6.txt","r") as data:
    # Read all lines into a list
    list_of_lines = data.readlines()

# Iterate through the list of lines and print even-numbered lines (1-based indexing)
# Python lists are 0-indexed, so the 2nd line is at index 1, 4th at index 3, etc.
# This means we start at index 1 and increment by 2.
for i in range(1, len(list_of_lines), 2):
    print(list_of_lines[i].strip()) # .strip() removes the newline character at the end of each line
````
